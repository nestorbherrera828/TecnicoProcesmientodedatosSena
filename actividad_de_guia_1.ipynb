```python
import pandas as pd

# Cargar el archivo de Excel
file_path = '/content/drive/MyDrive/Colab Notebooks/ACTIVIDAD GUIA 1/ips_Bogota_20260922.xlsx'
df = pd.read_excel(file_path)

# Mostrar información general del dataset
print("--- Información General del Dataset ---")
print(df.info())

# Mostrar las primeras filas
print("\n--- Primeras 5 filas del Dataset ---")
display(df.head())

# Mostrar estadísticas descriptivas de las columnas numéricas
print("\n--- Estadísticas Descriptivas ---")
display(df.describe(include='all'))
```

```python
# 1. Verificación y eliminación de filas duplicadas
duplicados_iniciales = df.duplicated().sum()
print(f"Filas duplicadas encontradas: {duplicados_iniciales}")
if duplicados_iniciales > 0:
    df = df.drop_duplicates()
    print("¡Duplicados eliminados!")

# 2. Limpieza de strings (eliminar espacios en blanco adicionales y estandarizar)
columnas_texto = ['ESE', 'NaturalezaJuridica', 'ClasePrestadorDesc']
for col in columnas_texto:
    if col in df.columns:
        # Quitar espacios al inicio/final y asegurar formato tipo título o mayúsculas
        df[col] = df[col].astype(str).str.strip()

# 3. Conversión y limpieza de tipos de datos
# Los códigos de prestador y habilitación son identificadores, es mejor tratarlos como texto (str)
if 'CodigoPrestador' in df.columns:
    df['CodigoPrestador'] = df['CodigoPrestador'].astype(str)
if 'CodigoHabilitacionSede' in df.columns:
    df['CodigoHabilitacionSede'] = df['CodigoHabilitacionSede'].astype(str)

# 4. Mostrar el dataset limpio
print("\n--- Información del Dataset después de la Limpieza ---")
print(df.info())
print("\n--- Primeras filas del Dataset limpio ---")
display(df.head())
```

```python
# 1. Identificar si existen valores nulos en el dataset
valores_nulos = df.isnull().sum()

print("--- Cantidad de valores nulos por columna ---")
print(valores_nulos)

# 2. Decisión de limpieza basada en los hallazgos
total_nulos = valores_nulos.sum()
if total_nulos == 0:
    print("\nNo se encontraron valores nulos en el dataset. ¡Tu base de datos está completa!")
else:
    print(f"\nSe encontraron {total_nulos} valores faltantes en total.")
    # Ejemplo opcional para eliminar filas con nulos si existieran:
    # df_limpio = df.dropna()
    # print("Filas con nulos eliminadas.")
```

```python
# Calcular cantidad y porcentaje de datos nulos por columna
nulos_por_columna = df.isnull().sum()
porcentaje_nulos = (df.isnull().sum() / len(df)) * 100

# Crear un DataFrame para mostrar el reporte resumido
reporte_nulos = pd.DataFrame({
    'Valores Nulos': nulos_por_columna,
    'Porcentaje (%)': porcentaje_nulos
})

print("--- Reporte de Datos Nulos en la Base de Datos ---")
display(reporte_nulos)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el estilo de los gráficos
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Gráfico para Naturaleza Jurídica
sns.countplot(
    data=df,
    x='NaturalezaJuridica',
    ax=axes[0],
    hue='NaturalezaJuridica',
    palette='viridis',
    order=df['NaturalezaJuridica'].value_counts().index,
    legend=False
)
axes[0].set_title('Distribución por Naturaleza Jurídica')
axes[0].set_xlabel('Naturaleza Jurídica')
axes[0].set_ylabel('Cantidad')
axes[0].tick_params(axis='x', rotation=45)

# 2. Gráfico para Clase de Prestador
sns.countplot(
    data=df,
    x='ClasePrestadorDesc',
    ax=axes[1],
    hue='ClasePrestadorDesc',
    palette='plasma',
    order=df['ClasePrestadorDesc'].value_counts().index,
    legend=False
)
axes[1].set_title('Distribución por Clase de Prestador')
axes[1].set_xlabel('Clase de Prestador')
axes[1].set_ylabel('Cantidad')
axes[1].tick_params(axis='x', rotation=45)

# 3. Gráfico para ESE (Empresa Social del Estado)
sns.countplot(
    data=df,
    x='ESE',
    ax=axes[2],
    hue='ESE',
    palette='coolwarm',
    order=df['ESE'].value_counts().index,
    legend=False
)
axes[2].set_title('Distribución de ESE')
axes[2].set_xlabel('¿Es ESE?')
axes[2].set_ylabel('Cantidad')

plt.tight_layout()
plt.show()
```
